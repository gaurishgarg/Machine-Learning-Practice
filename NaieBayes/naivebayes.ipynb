```python
test_example = {"Outlook":"Rainy","Temp":"Cool","Humidity":"High","Windy":'t'}
test_example
```

```python
def prediction(df,test_example):
    probabilties = {}
    targets = df["Play"].unique()
    for value in targets:
        probabilties[value] = (df["Play"][df["Play"]==value].count())/len(df)

    probs = {}
    for target in probabilties.keys():
        logsum = 0
        den = df["Play"][df["Play"]==target].count()
        for testvalue in test_example.keys():
            num = df["Play"][(df[testvalue]==test_example[testvalue]) & (df["Play"]==target)].count()
            independent = (num+1)/(den+4)
            logsum = logsum+np.log(independent)        
        logprob = np.log(probabilties[target])+logsum
        probs[target] = np.exp(logprob)
    max_key = max(probs, key=probs.get)
    return max_key
```

```python
prediction(df,test_example=test_example)
```
