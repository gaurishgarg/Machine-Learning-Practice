Importing Builtin Libraries

````python
import os
````

#### Loading a Corpus

Using File Method

````python
filenames = os.listdir("./Data")
````

````python
corpus = []
for filename in filenames:
    path = os.path.join("Data",filename)
    f = open(path,'r')
    text = f.read()
    corpus.append(text)
    f.close()
````

Using Inbuilt Libraries

````python
from nltk.corpus import PlaintextCorpusReader
dataset = PlaintextCorpusReader("./Data",".*")
````

Here `".*"` means match all filenames. If we want to match filenames starting with `doc`, we would use `"doc.*"`, and for matching only text files starting with `doc`, we will use `"doc.*\.txt"`.

**Therefore:**

- `".*"` → all files  
- `"doc.*"` → files starting with `"doc"`  
- `"doc.*\.txt"` → files starting with `"doc"` and ending with `".txt"`

````python
corpus = [dataset.raw(file) for file in dataset.fileids()]
````

#### Preprocessing

Normalisation

````python
alpha_corpus = []
punctuations = ['.', ',', '!', '?', ';', ':']
for document in corpus:
    document = ''.join(ch for ch in document if ch not in punctuations) ## join without space
    lower_words = [word.lower() for word in document.split()]
    alpha_words = [word for word in lower_words if word.isalpha()]
    normalised_document = ' '.join(alpha_words)
    alpha_corpus.append(normalised_document)
    
````

What does `' '.join(list)` mean?

It is used to **combine a list of strings into a single string**, with a space between each element.

Syntax: 
```python
'separator'.join(iterable)

Tokenization

````python
tokenised_corpus = []
for document in alpha_corpus:
    words = [word for word in document.split()]
    tokenised_corpus.append(words)
````

Stopword removal

````python
import nltk
nltk.download('stopwords')
````

````python
from nltk.corpus import stopwords
stop = stopwords.words(fileids='english')
````

````python
nostopword_corpus = []
for document in tokenised_corpus:
    nostopwords = [word for word in document if word not in stop]
    nostopword_corpus.append(nostopwords)
````

Stemming

````python
from nltk.stem import PorterStemmer
````

````python
ps = PorterStemmer()
stemmed_corpus = []
for document in nostopword_corpus:
    stemmed = [ps.stem(word) for word in document if word not in stop]
    stemmed_corpus.append(stemmed)
````

### Bag of Words

## Bag of Words (BoW)

**Definition:**  
Bag of Words is a text representation technique in Natural Language Processing (NLP) where a document is represented as a collection (bag) of words, ignoring grammar and word order but keeping track of word frequency.

- Each unique word in the corpus forms the **vocabulary**
- Each document is represented as a **vector of word counts**
- Word order is **not considered**

## Vocabulary:

[Word1, Word2, Word3, Word4]


### Document-Term Matrix:

|            | Word1 | Word2 | Word3 | Word4 |
|------------|------|------|------|------|
| Document 1 | 1    | 0    | 2    | 1    |
| Document 2 | 1    | 0    | 0    | 0    |
| Document 3 | 0    | 1    | 1    | 1    |


## Explanation

- **Document 1** contains:
  - Word1 → 1 time  
  - Word3 → 2 times  
  - Word4 → 1 time  

- **Document 2** contains:
  - Word1 → 1 time  

- **Document 3** contains:
  - Word2, Word3, Word4 → each 1 time  

---

## Limitations

- Ignores word order and context  
- Can create large sparse matrices  
- Does not capture semantics  

````python
vocabulary = []
for document in stemmed_corpus:
    vocabulary += document
unique_vocabulary = list(set(vocabulary)) ## set method gets unique vocabulary
````

````python
bow = []
i = 0
for document in stemmed_corpus:
    counts = {}
    document_id = i
    i += 1
    counts["document_id"] = document_id
    for word in unique_vocabulary:
        counts[word] = document.count(word)
    bow.append(counts)
````

````python
import pandas as pd
````

````python
bow_Df = pd.DataFrame(bow)
bow_Df
````

## TF-IDF (Term Frequency – Inverse Document Frequency)

**Definition:**  
TF-IDF is a numerical measure that indicates how important a word is to a document in a corpus. It is calculated by multiplying **Term Frequency (TF)** and **Inverse Document Frequency (IDF)**.

---

### Formula

- TF(t, d) = frequency of term *t* in document *d*  
- IDF(t) = log(N / df)  
  - N = total number of documents  
  - df = number of documents containing the term  

- **TF-IDF(t, d) = TF(t, d) × IDF(t)**

---

### Example

|        | Doc1_TF | Doc2_TF | Doc3_TF | IDF |
|--------|--------|--------|--------|-----|
| Word1  | 1      | 0      | 2      | 1   |
| Word2  | 1      | 0      | 0      | 0   |
| Word3  | 0      | 1      | 1      | 1   |

---

### TF-IDF Matrix

|        | Doc1_TFIDF | Doc2_TFIDF | Doc3_TFIDF |
|--------|-----------|-----------|-----------|
| Word1  | 1         | 0         | 2         |
| Word2  | 0         | 0         | 0         |
| Word3  | 0         | 1         | 1         |

---

### Interpretation

- Words with **high TF and high IDF** are most important  
- Words appearing in all documents get **low IDF (≈0)**  
- TF-IDF highlights **distinctive words in each document**

````python
import numpy as np
tf_idf = []
for word in vocabulary:
    features = {}
    features["Word"] = word
    i = 0
    d = 0
    n = len(stemmed_corpus)
    for document in stemmed_corpus:
        feature_name = "TF_"+str(i)
        features[feature_name] = (document.count(word)/len(document))
        i = i+1
        if(word in document):
            d = d+1
    idf = np.log(n/d)+1
    features["IDF"] = idf        
    i = 0
    for document in stemmed_corpus:
        term_Frequency_Name = "TF_"+str(i)
        new_Feature_Name = "TF-IDF_"+str(i)
        features[new_Feature_Name] = features[term_Frequency_Name]*idf
        i = i+1
    tf_idf.append(features)
````

````python
tf_idf_Df = pd.DataFrame(tf_idf)
tf_idf_Df
````

````python
docs = []
for i in range(0,len(stemmed_corpus)):
    word_scores = {}
    word_scores["DocumentID"] = "Document_"+str(i)
    for object in tf_idf:
        word_scores[object["Word"]] = object["TF-IDF_"+str(i)]
    docs.append(word_scores)
tf_idf_df_t = pd.DataFrame(docs)
tf_idf_df_t
````

## Manual TF-IDF vs Sklearn TF-IDF

**Observation:**  
Manual TF-IDF ≈ 0.15, while sklearn TF-IDF ≈ 3.3. However,  
`3.3 / 22 ≈ 0.15`, showing a clear relationship between the two.

**Explanation:**  
In the manual approach:  
`TF = count / total_words`, `IDF = log(n/d)`, so  
`TF-IDF = (1/22) × 3.3 ≈ 0.15`.

In sklearn:  
`TF = raw count`, `IDF ≈ 3.3`, so  
`TF-IDF = 1 × 3.3 = 3.3`.

**Key Relationship:**  
`Manual TF-IDF = Sklearn TF-IDF / document_length`

**Core Difference:**  
- Manual method normalizes TF by document length  
- Sklearn uses raw counts and may apply normalization later  

**Conclusion:**  
Both methods are mathematically correct. The difference arises from how **TF is defined (normalized vs raw count)**, not from an error in computation.
