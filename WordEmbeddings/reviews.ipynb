#### Making the Corpus

```python
df = df.drop(columns="sentiment",axis=1)
df.head()
```

```python
df = df.iloc[:1000,0]
df.info()
```

```python
corpus = list(df)
```

```python
len(corpus)
```

```python
alpha_corpus = []
punctuations = ['/','.', ',', '!', '?', ';', ':','\\']
for document in corpus:
    document = ''.join(ch for ch in document if ch not in punctuations) ## join without space
    lower_words = [word.lower() for word in document.split()]
    alpha_words = [word for word in lower_words if word.isalpha()]
    normalised_document = ' '.join(alpha_words)
    alpha_corpus.append(normalised_document)
```

```python
alpha_corpus[0]
```

```python
tokenised_corpus = []
for document in alpha_corpus:
    tokens = [word for word in document.split()]
    tokenised_corpus.append(tokens)
tokenised_corpus[0]
```

```python
from nltk.corpus import stopwords
```

```python
sw = stopwords.words(fileids='english')
sw
```

```python
nostop_corpus = []
for document in tokenised_corpus:
    nostop = [word for word in document if word not in sw]
    nostop_corpus.append(nostop)
nostop_corpus[0]
```

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()
```

```python
stemmed_corpus = []
for document in nostop_corpus:
    stemmed = [ps.stem(word) for word in document if word not in sw]
    stemmed_corpus.append(stemmed)
len(stemmed_corpus), len(stemmed_corpus[0])
```

```python
vocabulary = []
for document in stemmed_corpus:
    vocabulary += document
len(vocabulary)
```

```python
vocabulary = list(set(vocabulary))
len(vocabulary)
```

Converting the corpus into binary BOW vector of size mXn, where m=1000 (number of reviews 
documents) and n is the number of unique terms obtained from the 1000 documents. Each ij th entry 
of the vector is a binary value which is 1 if the jth term is present in ith review else 0. 

```python
bow = []
for document in stemmed_corpus:
    dic = {}
    for word in vocabulary:
        dic[word] = 1 if word in document else 0
    bow.append(dic)
```

```python
bow[0]["violenc"]
```

```python
matrix = {word: {w: 0 for w in vocabulary} for word in vocabulary}
matrix["violenc"]["violenc"]
```

```python
print(len(stemmed_corpus))
print(stemmed_corpus[:2])
```

```python
ws = 3
for document in stemmed_corpus:
    for i in range(len(document)):
       word = document[i]
       start = max(0, i - ws)
       end = min(len(document), i + ws + 1)

       for j in range(i-1, start-1, -1):
         matrix[word][document[j]] += 1

       for j in range(i+1, end):
          matrix[word][document[j]] += 1
```

```python
import numpy as np
mat = np.array(pd.DataFrame(matrix))
```

```python
from sklearn.decomposition import TruncatedSVD
svd = TruncatedSVD(n_components=100)
truncated = svd.fit_transform(mat)
```
