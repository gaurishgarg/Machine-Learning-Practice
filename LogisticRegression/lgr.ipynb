````python
import numpy as np
import pandas as pd
````

````python
from sklearn.datasets import load_iris
````

````python
temp =  load_iris()
temp
````

````python
df = pd.DataFrame(temp.data)
df.head()
````

````python
targets = pd.DataFrame(temp.target)
targets.head()
````

````python
df1 = pd.concat([df,targets],axis=1)
df1.head()
````

````python
df = df1.sample(frac=1)
df
````

````python
df.columns = ["f0", "f1", "f2", "f3", "t"]
````

````python
df
````

````python
df = pd.get_dummies(df,columns=["t"],dtype=int)
df.head()
````

````python
##For T0:
x = df.iloc[:,0:4]
y1 = df.iloc[:,4]
y2 = df.iloc[:,5]
y3 = df.iloc[:,6]
````

````python
len(x)
````

````python
##test-train-split via 70-30
train_limit = int(0.7*len(x))
train_limit
````

````python
x_Tr = x.iloc[:train_limit,:]
y1_Tr = y1.iloc[:train_limit]
y2_Tr = y2.iloc[:train_limit]
y3_Tr = y3.iloc[:train_limit]

len(x_Tr),len(y1_Tr),len(y2_Tr),len(y3_Tr)
````

````python
x_T = x.iloc[train_limit:len(x),:]
y1_T = y1.iloc[train_limit:len(y1)]
y2_T = y2.iloc[train_limit:len(y2)]
y3_T = y3.iloc[train_limit:len(y3)]

len(x_T),len(y1_T),len(y2_T),len(y3_T)
````

````python
#let alpha be 0.01, using k fold cross validation with k = 3
fold_size = int(len(x_Tr)/3)
fold_size

````

````python
##for y1,y2,y3
````

````python
folds = []
for i in range(0,len(x_Tr),fold_size):
    inputs = x_Tr.iloc[i:i+fold_size,:]
    y1_outputs = y1_Tr.iloc[i:i+fold_size]
    y2_outputs = y2_Tr.iloc[i:i+fold_size]
    y3_outputs = y3_Tr.iloc[i:i+fold_size]
    folds.append((inputs,y1_outputs,y2_outputs,y3_outputs))
````

````python
folds[0][0].head(),folds[0][1].head(),folds[0][2].head(),folds[0][3].head()
````

````python
from sklearn.metrics import roc_auc_score
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt
````

````python
alpha = 0.015
for yval in range(0,3):        
    for k in range(0,3):
        train_inputs = pd.concat([folds[i][0] for i in range(0,len(folds)) if i!=k],axis=0)
        train_outputs = pd.concat([folds[i][yval+1] for i in range(0,len(folds)) if i!=k],axis=0)
        validation_set_inputs = folds[k][0]
        validation_set_outputs = folds[k][yval+1]
        betas = np.zeros(x.shape[1]+1)    
        ##epochs
        total_epochs = 100000
        for epoch in range(0,total_epochs):
            x_train = np.c_[np.ones(len(train_inputs)),np.array(train_inputs)]
            xbybeta = np.matmul(x_train,betas)
            predictions = 1/(1+np.exp(-xbybeta))
            actuals = np.array(train_outputs).ravel()
            gradient = np.matmul(x_train.transpose(),predictions-actuals)
            n = len(x_train)
            betas = betas - 1/n*alpha*gradient
        ##validation
        x_val = np.c_[np.ones(len(validation_set_inputs)),np.array(validation_set_inputs)]
        xbybeta = np.matmul(x_val,betas)
        predictions = 1/(1+np.exp(-xbybeta))
        scores = [1 if value >= 0.5 else 0 for value in predictions]
        y_val = np.array(validation_set_outputs).ravel()
        aucrocscore = roc_auc_score(y_val,predictions)
        print(f"fold: {k} class: {yval} alpha: {alpha} total_epochs: {total_epochs}, score: {aucrocscore}")    
    
````

### Multi-Class Prediction using One-vs-Rest (OvR)

In this approach, we extend binary logistic regression to multi-class classification by training **one classifier per class**.

---

#### Step 1: Train Binary Classifiers

For each class ( k \in {0,1,2} ):

* Train a logistic regression model to distinguish:
  [
  y = k \quad \text{vs} \quad y \neq k
  ]

This results in **three sets of parameters (betas)**:

* ( \beta_0 ) → class 0 vs rest
* ( \beta_1 ) → class 1 vs rest
* ( \beta_2 ) → class 2 vs rest

---

#### Step 2: Compute Probabilities

For a given input ( x ), each classifier outputs a probability using the sigmoid function:

[
P(y = k \mid x) = \frac{1}{1 + e^{-\beta_k^T x}}
]

We compute this for all classes:

[
\text{probs} = [P(y=0|x),; P(y=1|x),; P(y=2|x)]
]

---

#### Step 3: Combine Outputs

The final prediction is obtained by selecting the class with the **highest probability**:

[
\hat{y} = \arg\max_k P(y = k \mid x)
]

---

#### Step 4: Implementation

```python
all_probs = []

for beta in final_betas:
    probs = 1 / (1 + np.exp(-np.matmul(x, beta)))
    all_probs.append(probs)

# Convert to shape (n_samples, n_classes)
all_probs = np.array(all_probs).T

# Final prediction
y_pred = np.argmax(all_probs, axis=1)
```

---

#### Important Notes

* The probabilities from different classifiers are **independent** and may not sum to 1.
* The `argmax` operation ensures we select the most confident class.
* This method is simple and effective, but differs from softmax regression where probabilities are normalized.

---

#### Summary

* Train ( K ) binary classifiers (one per class)
* Compute probability for each class
* Choose class with highest probability

[
\boxed{\hat{y} = \arg\max_k \sigma(\beta_k^T x)}
]
